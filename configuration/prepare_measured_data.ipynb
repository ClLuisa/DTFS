```python
#Dry bulb temperature, Total radiation on surface, Beam radiation on surface, Angle of incidence on surface, Sun zenith angle, Sun azimuth angle, Relative humidity, Wind speed, Effective sky temperature, Ground reflectance, Flap State Code

flap_state_map = {
    'ALL_CLOSED' : 0,
    'ALL_OPEN' : 2,
    'OUTBOUND_CLOSED_INBOUND_OPEN': 1,
    'OUTBOUND_OPEN_INBOUND_CLOSED': -1
}

prepared_data = pd.DataFrame()

prepared_data["Dry bulb temperature"] = measured_data["OUTDOOR_AIR_TEMP_C"]
prepared_data["Total radiation on surface"] = measured_data["SOLAR_RADIATION_VERTICAL_W_M2"]
prepared_data["Beam radiation on surface"] = measured_data["SOLAR_RADIATION_VERTICAL_W_M2"]
prepared_data["Angle of incidence on surface"] = 90
prepared_data["Sun zenith angle"] = 89
prepared_data["Sun azimuth angle"] = 180
prepared_data["Relative humidity"] = measured_data["OUTDOOR_RELATIVE_HUMIDITY_PERCENT"]
prepared_data["Wind speed"] = 0
prepared_data["Effective sky temperature"] = measured_data["OUTDOOR_AIR_TEMP_C"]
prepared_data["Ground reflectance"] = 0.2
prepared_data["Flap State Code"] = measured_data["logical_flap_state"].map(flap_state_map)
prepared_data["Indoor Air Temp"] = measured_data["INDOOR_AIR_TEMP_ANALOG_C"]
prepared_data['Outlet Air Temperature'] = measured_data['TOP_INBOUND_AIR_TEMP_1_C']
prepared_data['Wall Temperature'] = measured_data['PCM_OC25_TEMP_C']
```

```python
# Write header manually (comma + space separated), then data (tab separated)
with open(r"configuration\prepared_data.txt", "w") as f:
    f.write(", ".join(prepared_data.columns) + "\n")
    prepared_data.to_csv(f, sep="\t", header=False, index=False, lineterminator="\n")
```
